```python
area = pd.read_csv("area_protegida.csv")
```

**1. ¿Cual es el tamaño total de la superficie protegida y cual es el promedio de cada area?**

si queremos saber cuanta tierra y agua esta protegida en total en el pais y cuanto mide en promedio una de estas areas podriamos usar el siguiente codigo.

```python
area_total = area["area"].sum()
area_promedio = area["area"].mean()

print("superficie total protegida:", area_total)
print("superficie promedio de cada area:", area_promedio)
```

**2. ¿Cuales son las 5 categorias de proteccion que mas superficie protegen?**

en si hay categorias como "Parque Nacional" o "Reserva Provincial" asi que con el siguiente codigo podremos ver cuales de estas categorias suman mas terreno protegido en total.

```python
top_categorias_area = area.groupby("gna")["area"].sum().nlargest(5)
print(top_categorias_area)
```

**3. ¿Cuantas areas protegidas tienen un tamaño menor a la media?**

la mitad de las areas del pais estan por debajo de la media en tamaño (79,300) con este codigo podremos contar cuantas son exactamente.

```python
mediana_area = area["area"].median()
areas_pequenas = area[area["area"] < mediana_area]

print("la cantidad de areas mas pequeñas que la mitad de la tabla son:", len(areas_pequenas))
```

**4. ¿Que porcentaje de las areas no tiene cargado el documento legal (fdc)?**

Si nosotros quisieramos saber cuantas areas están "olvidadas" en el registro y no tienen cargada la ley o decreto que las creo para ver su legalidad actual podriamos usar el siguiente comando.

```python
porcentaje_sin_ley = (area["fdc"].isna().mean() * 100).round(2)
print("porcentaje de areas sin el documento legal:", porcentaje_sin_ley, "%")
```

**5. ¿Cual es la reserva mas pequeña de todas y cuanto mide?**

este codigo es por si queremos encontrar el area protegida mas chiquita de todo el mapa.

```python
mas_chica = area.nsmallest(1, "area")[["nam", "gna", "area"]]
print(mas_chica)
```

**6. ¿Cuantas "areas marinas protegidas" hay y cuanto territorio suman?**

con este codigo queremos ver cuantas de estas areas protegen especificamente el mar y el agua y que tan grandes son juntas.

```python
marinas = area[area["gna"] == "area marina protegida"]
print("cantidad de areas marinas:", len(marinas))
print("suma de su superficie marina:", marinas["area"].sum())
```

**7. ¿Cuantas areas maneja el instituto geografico nacional (IGN) frente al ministerio de ambiente (MAyDS)?**

si necesitamos ver el conteo de cuantas areas tiene bajo su control cada organismo del gobierno este codigo nos lo muestra.

```python
conteo_organismos = area["sag"].value_counts()
print(conteo_organismos)
```

**8. ¿Cuales son los nombres de las 3 areas con mayor superficie de la muestra?**

aca esta una lista rapida con los nombres de los 3 "gigantes" del mapa.

```python
top3_nombres = area.nlargest(3, "area")[["nam", "area"]]
print(top3_nombres)
```

**9. ¿donde estan ubicados los "parques nacionales"?**

en el mapa se ve claramente cada punto del pais en donde se puede ver el nombre de cada parque nacional.

```python
import folium

parques = area[area["gna"] == "Parque Nacional"]
mapa_parques = folium.Map(location=[-38.41, -63.61], zoom_start=4)

for _, fila in parques.iterrows():
    folium.Marker(
        location=[fila["lat"], fila["lng"]],
        popup=fila["nam"]
    ).add_to(mapa_parques)
    
mapa_parques.save("mapa_parques.html")
mapa_parques
```

**10. ¿donde estan las 15 areas protegidas mas grandes?**

en el siguiente mapa se podran ver las ubicaciones de las 15 areas protegidas mas grandes.

```python
import folium

grandes = area.nlargest(15, "area")
mapa_grandes = folium.Map(location=[-38.41, -63.61], zoom_start=4)

for _, fila in grandes.iterrows():
    folium.CircleMarker(
        location=[fila["lat"], fila["lng"]],
        radius=12,
        color="red",
        fill=True,
        fill_opacity=0.6,
        popup=f"{fila['nam']} (Tamaño: {fila['area']})"
    ).add_to(mapa_grandes)

mapa_grandes.save("mapa_grandes.html")
mapa_grandes
```
